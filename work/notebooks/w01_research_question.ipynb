## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import duckdb
import pandas as pd
from google.colab import userdata

# Connect to DuckDB
con = duckdb.connect()

# Set Hugging Face secret token from Colab Secrets
hf_token = os.getenv("HF_TOKEN", "")
if not hf_token:
    try:
        hf_token = userdata.get('HF_TOKEN')
        os.environ["HF_TOKEN"] = hf_token
    except userdata.SecretNotFoundError:
        print("HF_TOKEN not found in Colab secrets.")

if hf_token:
    con.execute(f"CREATE OR REPLACE SECRET hf (TYPE huggingface, TOKEN '{hf_token}')")

# Data repository URL
REL = 'hf://datasets/FlyRank/internship-warehouse'

# Query 90-day query performance from fact_content_query_90d using exact column names
query_sql = f"""
SELECT
    COUNT(DISTINCT query_hash_id) AS total_unique_queries,
    COUNT(*) AS total_records,
    ROUND(AVG(clicks_90d), 2) AS avg_clicks,
    ROUND(AVG(content_total_impressions_90d), 2) AS avg_impressions
FROM read_parquet('{REL}/fact_content_query_90d.parquet')
"""

df_stats = con.sql(query_sql).df()

print("=== LANE 2 DATASET METRICS (fact_content_query_90d) ===")
print(f"1. Total Unique Queries:    {df_stats['total_unique_queries'][0]:,}")
print(f"2. Total Records:           {df_stats['total_records'][0]:,}")
print(f"3. Average 90D Clicks:      {df_stats['avg_clicks'][0]:,}")
print(f"4. Average 90D Impressions: {df_stats['avg_impressions'][0]:,}")
```
